<font color='#4393C3' size='5' style="text-decoration: underline; text-decoration-color: blue;"><b>Analysis of Missing Data</b></font>

To understand the potential impact of missing values, we can compare the summary statistics of rows with missing data to those without. This comparison will help us determine whether the missing data is completely at random (MCAR) or if there is an underlying pattern causing it.

```python
# filtering through the rows with NaN values in the 'label' column
null_mask = waze_data['label'].isnull()
waze_data_null = waze_data[null_mask]
waze_data_null.describe()
```

```python
# filtering through the rows without NaN values in the 'label' column
not_null_mask = ~waze_data['label'].isnull()
waze_data_not_null = waze_data[not_null_mask]
waze_data_not_null.describe()
```

By comparing the two summary statistics, especially focusing on median values and standard deviations, we can see that the results are consistent for both datasets (with and without missing values). Therefore, we can assume that part of the data is missing for random reasons, suggesting that the missing values are likely Missing Completely at Random (MCAR).

<font color='#4393C3' size='5' style="text-decoration: underline; text-decoration-color: blue;"><b>If Missing Data Depends on the User's Device Type</b></font>

We have two types of devices in use: Android and iPhone. If the missing data can be explained by a user's device type, we can use the already filtered waze_data_null dataframe to examine how missing values are distributed between Android and iPhone devices.

```python
# Proportions of the device types in rows with missing labels 
waze_data_null['device'].value_counts(normalize = True).reset_index()
```

```python
# Proportions of the device types in the dataset overall 
waze_data['device'].value_counts(normalize = True).reset_index()
```

We observe approximately the same proportion of devices in the overall dataset and in the dataset with missing 'label' info: 447 iPhone (~ **64%**) vs. 253 (~ **36%**) Android devices. We can assume that the presence of missing values is not dependent on device type.

<font color='#4393C3' size='5' style="text-decoration: underline; text-decoration-color: blue;"><b>Churned vs. Retained Users</b></font>

Now, we can proceed with the analysis of the proportion of churned vs. retained users.

```python
# Proportions of churned and retained uesers in the dataset
waze_data_not_null['label'].value_counts(normalize = True).reset_index()
```

There is 82% of retained users and 18% churned users in the dataset. 

```python
# Descriptive statistics for the churned users
churned = waze_data_not_null[waze_data_not_null['label'] == 'churned']
churned.describe()
```

```python
# Descriptive statistics for the retained users
retained = waze_data_not_null[waze_data_not_null['label'] == 'retained']
retained.describe()
```

Let's compare the medians (which is better than the mean, because we don’t yet know if our data might be skewed).

```python
# Median values calculation for both 'populations'
waze_data.groupby('label').median(numeric_only=True).reset_index()
```

The median values show that churned users have 59 sessions, while retained users have 56 sessions during the last month. We can also see that churned users drove ~200 more kilometers and more that two hours han retained users, even though they have approximately half as much activity (just log-ins) and fewer actual driving days.

```python
# Create a new column 'distance_per_drive' 
waze_data['distance_per_drive'] = waze_data['driven_km_drives'] / waze_data['drives']

# The calculation of the median distance per drive for both 'populations'
waze_data.groupby('label')['distance_per_drive'].median(numeric_only=True).reset_index()
```

The median distance per drive for retained users is slightly higher than that for churned users.

```python
# Create a new column 'km_per_driving_day'. Some values in this column will be 'inf' 
# because a drive is only recorded if it exceeds 1 km. If any drive a driver makes  
# over the month is less than 1 km, then the 'drives' column remains 0 while other columns 
# are still updated, leading to division by zero.
waze_data['km_per_driving_day'] = waze_data['driven_km_drives'] / waze_data['driving_days']

# The calculation of the median distance driven per day for both 'populations'
waze_data.groupby('label')['km_per_driving_day'].median(numeric_only=True).reset_index()
```

```python
# How many drives per day each group made
waze_data['drives_per_day'] = waze_data['drives']/waze_data['driving_days']

# Groupping and calculate the median value
waze_data.groupby('label')['drives_per_day'].median(numeric_only = True).reset_index()
```

From our calculations, we can see that an average churned driver drove approximately 698 km per day, which is about 2.4 times more than an average retained user. Moreover, churned users made 10 drives per day on average, while retained users made around 4. We can assume that this disproportion is due to churned users being long-distance truck drivers (or in a similar profession), and perhaps the app does not meet their needs, leading them to stop using it.

<font color='#4393C3' size='5' style="text-decoration: underline; text-decoration-color: blue;"><b>Is the Proportion of Churned vs. Retained Users Related to Device Type?</b></font>

```python
# Absolute value counts for churned vs. retained users using iPhone or Android
waze_data.groupby(['label','device']).size().reset_index(name='count').sort_values(by='label')
```

```python
# Absolute value counts for churned vs. retained users using iPhone or Android
waze_data.groupby('label')['device'].value_counts(normalize=True).reset_index(name = 'proportion').sort_values(by='label')
```

The proportion of iPhone to Android users is similar in both the churned and retained groups, and it matches the overall proportion in the entire dataset.
